# 03. Collaborative Filtering Experiments
Tests TruncatedSVD matrix factorization on the user-item interaction matrix.

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import TruncatedSVD

df_interactions = pd.read_csv('../data/processed/interactions.csv')
pivot = df_interactions.pivot_table(index='learner_id', columns='course_id', values='rating').fillna(3.0)

svd = TruncatedSVD(n_components=10, random_state=42)
u_sigma = svd.fit_transform(pivot)
vt = svd.components_
reconstructed = np.dot(u_sigma, vt)
print("Reconstructed rating matrix shape:", reconstructed.shape)
print("Explained variance ratio sum:", svd.explained_variance_ratio_.sum())
```
