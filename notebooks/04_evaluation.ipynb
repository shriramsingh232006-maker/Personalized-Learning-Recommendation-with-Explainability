# 04. Offline Recommender Evaluation
Computes Precision@K, Recall@K, NDCG@K, and MAP@K on test learner interactions.

```python
import sys
from pathlib import Path
sys.path.append('..')

from db.database import SessionLocal
from src.recommenders.hybrid_engine import HybridEngine
from src.evaluation.metrics import RecommendationEvaluator

db = SessionLocal()
engine = HybridEngine()

for k in [3, 5, 10]:
    res = RecommendationEvaluator.evaluate_offline(db, engine, k=k)
    print(f"--- K = {k} ---")
    print(f"Precision@{k}: {res['precision_at_k']:.4f}")
    print(f"Recall@{k}:    {res['recall_at_k']:.4f}")
    print(f"NDCG@{k}:      {res['ndcg_at_k']:.4f}")
    print(f"MAP@{k}:       {res['map_at_k']:.4f}")

db.close()
```
