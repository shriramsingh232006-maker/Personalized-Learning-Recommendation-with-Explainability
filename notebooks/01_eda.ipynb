# 01. Exploratory Data Analysis (EDA)
## Personalized Learning Recommendation System
This notebook analyzes learner distributions, course topics, difficulty levels, and interaction patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

data_dir = Path('../data/processed')
df_learners = pd.read_csv(data_dir / 'learners.csv')
df_courses = pd.read_csv(data_dir / 'courses.csv')
df_interactions = pd.read_csv(data_dir / 'interactions.csv')

print(f"Learners: {len(df_learners)}")
print(f"Courses: {len(df_courses)}")
print(f"Interactions: {len(df_interactions)}")
```

```python
# Course topic breakdown
df_courses['topic'].value_counts().plot(kind='bar', title='Courses by Topic', color='teal')
plt.xlabel('Topic')
plt.ylabel('Count')
plt.show()
```

```python
# Interaction ratings & quiz score distribution
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df_interactions['rating'], bins=10, kde=True, ax=ax[0], color='orange')
ax[0].set_title('Interaction Ratings Distribution')

sns.histplot(df_interactions['quiz_score'], bins=15, kde=True, ax=ax[1], color='purple')
ax[1].set_title('Quiz Scores Distribution')
plt.show()
```
